```python
import pandas as pd
```

## Cleaning orders.csv

```python
df = pd.read_csv("orders.csv")
```

```python
df.head()
```

#### Fix Date Formats

```python
df["order_date"] = pd.to_datetime(df["order_date"],format="mixed",dayfirst=True)
```

```python
df.head()
```

#### Handle Missing Customer IDs and Correct datatype

```python
df["customer_id"] = df["customer_id"].fillna(0)
df["customer_id"] = df["customer_id"].astype("Int64")
```

```python
df
```

```python
df.to_csv("clean_orders.csv",index=False)
```

## Cleaning products.csv

```python
df = pd.read_csv("products.csv")
```

```python
df.head(10)
```

#### Remove Extra Spaces

```python
df["product_name"] = df["product_name"].str.strip()
```

```python
df.head(10)
```

#### Convert to Title Case

```python
df["product_name"] = df["product_name"].str.title()
```

```python
df.head(10)
```

```python
df.to_csv("clean_products.csv",index=False)
```

## validate emails

```python
df = pd.read_csv("customers.csv")
df.head()
```

```python
email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"
is_valid = df["email"].astype(str).str.match(email_pattern)
```

```python
bad_customer_ids = df.loc[~is_valid, "customer_id"].tolist()
 
df["email_valid"] = is_valid
df.to_csv("customers_clean.csv", index=False)
```

## check referential integrity

```python
orders = pd.read_csv("orders.csv")
items = pd.read_csv("order_items.csv")
```
