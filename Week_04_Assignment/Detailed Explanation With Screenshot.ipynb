### Task 1:

--> Create a Resource Group  

A Resource Group is a logical container in Microsoft Azure that stores and manages related cloud resources. Before creating any Azure service such as Storage Accounts or Azure Data Factory, a Resource Group must be created. It helps organize resources, monitor costs, apply permissions, and delete all associated resources together.

 , 

### Resource Groups provide logical organization for Azure resources. They allow developers and administrators to:

Manage multiple resources together.
Apply security permissions collectively.
Monitor billing for related resources.
Delete an entire project without deleting resources individually.

Azure Data Factory (ADF) is Microsoft's cloud-based ETL/ELT service used for creating automated data pipelines. This task focuses on understanding the ADF interface and connecting Azure Storage with Data Factory.

Azure Data Factory acts as an orchestration service that connects different data sources, transforms data, and automates workflows.

### Create Linked Service

A Linked Service acts like a connection string between Azure Data Factory and external resources.

Without a Linked Service, ADF cannot access data.

#### Why Linked Service?

It securely stores authentication details required to access data sources.

It acts similarly to a database connection in SQL applications.

### Create Datasets 
Datasets define the structure and location of the data used by Azure Data Factory.Datasets separate storage connection from actual data location, making pipelines reusable and easier to maintain.

#### They tell ADF:

Where data exists.

What format it is in.

Which file should be accessed.

Source Dataset --> CSV file stored inside Blob Storage.

Destination Dataset--> Target location where copied data will be stored.


### Get Metadata Activity

Get Metadata retrieves information about files without reading the actual content.It allows validation before processing.

Metadata Retrieved:

    File name
    Last modified date
    File size
    Child items
    Exists
    Structure

Azure uses Role-Based Access Control (RBAC) to securely manage permissions. This task demonstrates how to assign appropriate access rights to users and services.
### Reader Role --> Purpose:Provides read-only access.

#### Capabilities:
    View resources
    Monitor resources
    Cannot modify or delete

### Contributor Role --> Purpose: Allows users to create, modify, and delete resources.

#### Capabilities:

    Full resource management
    Cannot assign roles or manage access
